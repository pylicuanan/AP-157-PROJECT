```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

```python
heads = np.ones((5,8))
tails = np.ones((5,8))
tails[4,:] = 0
tails[4,2:4] = 1

xticks = np.arange(0,9)
yticks = np.arange(0,5)

extent = 0,8 , 0,5

heads_img = np.asarray(Image.open("grid graph/heads_grid.png"))
tails_img = np.asarray(Image.open("grid graph/tails_grid.png"))

img = plt.imshow(tails)
plt.close()

fig, ax = plt.subplots(2,1,sharex=True)
ax[0].imshow(heads,origin='lower', cmap='viridis_r', extent=extent)
ax[0].imshow(heads_img, origin='lower', extent=extent, alpha=0.5)
ax[0].set_title('Heads detection grid')
ax[0].grid('on')
ax[0].set_xticks(xticks)

ax[1].imshow(tails,origin='lower', cmap='viridis', extent=extent)
ax[1].imshow(tails_img, extent=extent, alpha=0.5)
ax[1].set_title('Tails detection grid')
ax[1].grid('on')
ax[1].set_xticks(xticks)
ax[1].set_xlabel('Number of coin widths (unit)')

fig.text(0.28, 0.5, 'Number of coin heights (unit)', va='center', rotation='vertical')
fig.colorbar(img, ax=ax, orientation='vertical', fraction=.1)

plt.savefig('grid graph/grid.png')
plt.show()
```
