```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

Set = np.array((2,4,6,8,10,12,14,16))
```

```python
Heads_Accuracy = []
Tails_Accuracy = []
for i in Set:
    file_path = fr"C:\Users\phili\OneDrive\Documents\UPD SY 2024-2025\2nd Sem\AP 157\PROJECT\PROJECT\datasets\num coins vs accuracy\{i}.csv"
    Data = pd.read_csv(file_path)
    heads_exp = Data.iloc[1:,1]
    tails_exp = Data.iloc[1:,2]
    heads_err = Data.iloc[1:,4]
    tails_err = Data.iloc[1:,5]
    heads_true = Data.iloc[1:,6]
    tails_true = Data.iloc[1:,7]
    heads_acc = np.mean((heads_exp - heads_err)/heads_true)
    tails_acc = np.mean((tails_exp - tails_err)/tails_true)
    Heads_Accuracy.append(heads_acc)
    Tails_Accuracy.append(tails_acc)
```

```python
plt.plot(Set, Heads_Accuracy, marker="o", label="Head Count Accuracy")
plt.plot(Set, Tails_Accuracy, marker="o", label="Tails Count Accuracy")
plt.ylim(0.0,1.1)
plt.xlabel("Number of Coins", fontsize=18)
plt.ylabel("Accuracy", fontsize=18)
plt.legend(fontsize=13)
plt.xticks(fontsize=15)
plt.yticks(fontsize=15)
plt.title("Heads/Tails detection accuracy for \n various number of Coins", fontsize=22)
plt.tight_layout()
plt.grid(True)
plt.savefig("graph.png")
```
